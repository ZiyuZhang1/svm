```python
file_path = "/itf-fi-ml/shared/users/ziyuzh/svm/results/2019_mf_add_pred/ICD10_D57_pred.pkl"

with open(file_path, "rb") as f:
    data = pickle.load(f)
```

```python
data.keys()
```

```python

y_true = np.array(data['true_label'])
y_pred = np.array(data['uniport_seq'])

# Step 1: sort indices by prediction score (descending)
sorted_indices = np.argsort(-y_pred)

# Step 2: create rank array
ranks = np.empty_like(sorted_indices)
ranks[sorted_indices] = np.arange(1, len(y_pred) + 1)

# Step 3: get ranks of positive samples
positive_ranks = ranks[y_true == 1]

print(positive_ranks)
```
