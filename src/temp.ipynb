```python

cols_to_keep = [
    'setting', 'method', 'category', 'bio_sim_300', 'bio_sim_150',
    'top_recall_150', 'succ_150', 'top_recall_300', 'succ_300',
    'bedroc_150', 'bedroc_300', 'auroc'
]

filtered_df = result_df[
    (result_df['setting'] == 'df_gnn_occsvm_pred') &
    (result_df['method'] == 'ppi_emb_dw')
][cols_to_keep]

result_mean = filtered_df.groupby('category').mean(numeric_only=True)
result_mean = result_mean.reset_index()
result_mean['method'] = 'SVM'
result_mean['feature'] = 'ppi_emb_dw'
```

```python
# Concatenate row-wise
combined_df = pd.concat([combined_df, result_mean], axis=0)

combined_df
```

```python
metrics = ['top_recall_150', 'top_recall_300', 'bedroc_150', 'bedroc_300']

ppi_better_categories = []

for method, method_df in combined_df.groupby('method'):
    ppi_df = method_df[method_df['feature'].str.startswith('ppi', na=False)]
    other_df = method_df[~method_df['feature'].str.startswith('ppi', na=False)]

    for category in method_df['category'].unique():
        ppi_cat = ppi_df[ppi_df['category'] == category]
        other_cat = other_df[other_df['category'] == category]

        if ppi_cat.empty or other_cat.empty:
            continue

        ppi_max = ppi_cat[metrics].max()
        other_max = other_cat[metrics].max()

        better_metrics = [
            metric for metric in metrics
            if ppi_max[metric] > other_max[metric]
        ]

        if better_metrics:
            ppi_better_categories.append({
                'method': method,
                'category': category,
                'better_metrics': better_metrics
            })

ppi_better_categories_df = pd.DataFrame(ppi_better_categories)

ppi_better_categories_df
```

```python
metrics = ['top_recall_150', 'top_recall_300', 'bedroc_150', 'bedroc_300']

non_ppi_better_categories = []

for method, method_df in combined_df.groupby('method'):
    non_ppi_df = method_df[~method_df['feature'].str.contains('ppi', na=False)]
    other_df = method_df[method_df['feature'].str.contains('ppi', na=False)]

    for category in method_df['category'].unique():
        non_ppi_cat = non_ppi_df[non_ppi_df['category'] == category]
        other_cat = other_df[other_df['category'] == category]

        if non_ppi_cat.empty or other_cat.empty:
            continue

        non_ppi_max = non_ppi_cat[metrics].max()
        other_max = other_cat[metrics].max()

        better_metrics = [
            metric for metric in metrics
            if non_ppi_max[metric] > other_max[metric]
        ]

        if better_metrics:
            non_ppi_better_categories.append({
                'method': method,
                'category': category,
                'better_metrics': better_metrics
            })

non_ppi_better_categories_df = pd.DataFrame(non_ppi_better_categories)

non_ppi_better_categories_df
```
