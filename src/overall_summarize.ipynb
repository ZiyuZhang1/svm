```python
import matplotlib.pyplot as plt
import numpy as np
def visualize_model_performance(model_sum_list,selected_disease, metric):
    df = model_sum_list[model_sum_list['disease'].isin(selected_disease)]

    diseases = df['disease'].unique()
    paras = sorted(df['para'].unique())

    n_dis = len(diseases)
    n_para = len(paras)

    x = np.arange(n_dis)     # positions = diseases
    bar_width = 0.15

    plt.figure(figsize=(12, 6))

    for i, p in enumerate(paras):
        sub = df[df['para'] == p].set_index('disease').loc[diseases]
        plt.bar(x + i * bar_width, sub[metric], width=bar_width, label=f"para={p}")

    plt.xticks(x + bar_width * (n_para - 1) / 2, diseases, rotation=45, ha='right')
    plt.ylabel(metric)
    plt.xlabel("Disease")
    plt.title(metric+" across diseases for different paras")
    plt.legend(title="para")
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def plot_overlap_heatmap(data, k=100,
                         features=['uniport_ppi_2019', 'ppi_2019_dw_40', 'diffusion_2019',
                                   'uniport_bio', 'uniport_seq', 'uniport_esm',
                                   'linear_fused', 'geo_fused']):
    """
    Compute top-k genes for each feature, calculate Jaccard similarity,
    and plot a heatmap of overlap.
    
    Parameters
    ----------
    data : pandas.DataFrame
        Must include score columns for features and a column 'test_genes'.
    k : int
        Number of top genes to select per feature.
    features : list of str
        Column names in data corresponding to prediction scores.
    """
    
    # Extract top-k predictions for each feature
    top_predictions = {}
    for feature in features:
        indices = np.argpartition(data[feature], -k)[-k:]
        indices = indices[np.argsort(data[feature][indices])[::-1]]
        top_predictions[feature] = set(data['test_genes'][indices])
    
    # Jaccard index
    def jaccard(set_a, set_b):
        return len(set_a & set_b) / len(set_a | set_b)
    
    # Compute similarity matrix
    labels = list(top_predictions.keys())
    matrix = np.zeros((len(labels), len(labels)))

    for i in range(len(labels)):
        for j in range(len(labels)):
            matrix[i, j] = jaccard(top_predictions[labels[i]], top_predictions[labels[j]])

    # Convert to DataFrame
    df = pd.DataFrame(matrix, index=labels, columns=labels)
    
    # Plot heatmap
    plt.figure(figsize=(8, 6))
    sns.heatmap(df, annot=True, fmt=".2f",
                cmap="bwr",  # blue-white-red
                vmin=0, vmax=1, center=0.5,  # control color scale
                square=True,
                cbar_kws={"label": "Jaccard Similarity"})
    plt.title(f"Top-{k} Gene Set Overlap Heatmap (Jaccard Index)")
    plt.tight_layout()
    plt.show()


```
