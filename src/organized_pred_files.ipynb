```python
root = '/itf-fi-ml/shared/users/ziyuzh/svm/results'
folders = ['df_early','df_early_ppi','2019_df_oob_lf','2019_df_oob_mid','2019_df_oob_ppi_mid','2019_occ_val_unlabled','2019_gnn_improved','2019_gnn_improved_sage']
out_path_pred = '/itf-fi-ml/shared/users/ziyuzh/svm/results/df_gnn_occsvm_pred'
os.makedirs(out_path_pred, exist_ok=True)
for filename in os.listdir('/itf-fi-ml/shared/users/ziyuzh/svm/results/df_early_pred/pred.pkl'):
    summarized_dict = dict()

    for idx, folder in enumerate(folders):
        folder = folder+'_pred'
        root_path = os.path.join(root, folder)
        if 'pred.pkl' in os.listdir(root_path):
            root_path = os.path.join(root_path,'pred.pkl')

        with open(os.path.join(root_path,filename), "rb") as f:
            data = pickle.load(f)

        if folders[idx] == 'df_early':
            summarized_dict['true_label'] = data['true_label'].copy()
            summarized_dict['test_genes'] = data['test_genes'].copy()
            summarized_dict['train_pos_genes'] = data['train_pos_genes'].copy()
            summarized_dict['early_fused'] = data['df_early'].copy()
        if folders[idx] == 'df_early_ppi':
            summarized_dict['early_fused_ppi'] = data['df_early_ppi'].copy()
        if folders[idx] == '2019_df_oob_lf':
            summarized_dict['late_fused'] = data['later_feature_fused_avg'].copy()
            summarized_dict['late_fused_ppi'] = data['later_ppi_feature_fused_avg'].copy()
        if folders[idx] == '2019_df_oob_mid':
            summarized_dict['ppi_emb_n2v'] = data['uniport_ppi_2019'].copy()
            summarized_dict['ppi_emb_dw'] = data['ppi_2019_dw_40'].copy()
            summarized_dict['ppi_emb_df'] = data['diffusion_2019'].copy()
            summarized_dict['literature_emb'] = data['uniport_bio'].copy()
            summarized_dict['seq_emb_esm'] = data['uniport_esm'].copy()
            summarized_dict['seq_emb_port'] = data['uniport_seq'].copy()

            summarized_dict['mid_linear_fused'] = data['linear_fused'].copy()
            summarized_dict['mid_geo_fused'] = data['geo_fused'].copy()       
        if folders[idx] == '2019_df_oob_ppi_mid':
            summarized_dict['mid_linear_fused_ppi'] = data['linear_fused'].copy()
            summarized_dict['mid_geo_fused_ppi'] = data['geo_fused'].copy()       
        if folders[idx] == '2019_occ_val_unlabled':
            summarized_dict['occ_svm'] = data['uniport_ppi_2019'].copy()         
        if folders[idx] == '2019_gnn_improved':
            summarized_dict['GCNs'] = data['gcn'].copy()  
        if folders[idx] == '2019_gnn_improved_sage':
            summarized_dict['Graph_sage'] = data['sage'].copy()    
    with open(os.path.join(out_path_pred, filename), 'wb') as f:
        pickle.dump(summarized_dict, f)
```
