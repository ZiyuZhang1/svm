```python
import pandas as pd
import os
```

```python
root = '/itf-fi-ml/shared/users/ziyuzh/svm/results'
files = ['2019_nn_oob','2019_df_oob','2019_gnn_improved','2019_rf_all']
```

```python
collection = []
for file in files:
    file_path = os.path.join(root,file+'.xlsx')
    df = pd.read_excel(file_path, sheet_name='disease')
    bad_disease_dict = dict()
    for disease in df['disease'].unique():
        subdf = df[df['disease']==disease]
        # if len(subdf[subdf['auroc']<=0.5]) >= 0.5*len(subdf):
        bad_subdf = subdf[subdf['auroc']<=0.5]
        if len(bad_subdf) >= 1:
            # print(disease)
            bad_disease_dict[disease]=bad_subdf['para'].tolist()
    collection.append(bad_disease_dict)
```

```python
pair_collection = []
n = 0
for file in files:
    file_path = os.path.join(root, file + '.xlsx')
    df = pd.read_excel(file_path, sheet_name='disease')

    for disease in df['disease'].unique():
        subdf = df[df['disease'] == disease]
        n+=len(subdf)

        # Find bad models (auroc <= 0.5)
        bad_subdf = subdf[subdf['auroc'] <= 0.5]
        if len(bad_subdf) >= 1:
            bad_model_lst = bad_subdf['para'].tolist()

            # --- Clean and modify model names ---
            cleaned_models = []
            for model in bad_model_lst:
                # Replace 'gcn' with 'uniport_ppi_2019'
                model = model.replace('gcn', 'uniport_ppi_2019')
                # Remove 'DL_' and 'RF_'
                model = model.replace('DL_', '').replace('RF_', '')
                cleaned_models.append(model)

            # Add pairs to the collection
            for model in cleaned_models:
                pair_collection.append((disease, model))
```
