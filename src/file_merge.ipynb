```python
import os
import pandas as pd

root1 = '/itf-fi-ml/shared/users/ziyuzh/svm/results/2019_nn_6_features'
root2 = '/itf-fi-ml/shared/users/ziyuzh/svm/results/2019_nn_3_features'
out_folder = '/itf-fi-ml/shared/users/ziyuzh/svm/results/2019_nn_merged'

# Ensure output dir exists
os.makedirs(out_folder, exist_ok=True)

for file_name in os.listdir(root1):
    if 'all' in file_name:
        continue

    file_path1 = os.path.join(root1, file_name)
    file_path2 = os.path.join(root2, file_name)

    # Skip if the counterpart is missing in root2
    if not os.path.isfile(file_path1) or not os.path.isfile(file_path2):
        continue

    df1 = pd.read_csv(file_path1)
    df2 = pd.read_csv(file_path2)
    df2['para'] = df2['para'].replace({
        "DL_early-0-0-0": "DL_early_ppi-0-0-0",
        "DL_mid-0-0-0": "DL_mid_ppi-0-0-0",
        "DL_later_avg-0-0-0": "DL_later_avg_ppi-0-0-0",
        "DL_later_mlp-0-0-0": "DL_later_mlp_ppi-0-0-0"
    })

    df2 = df2[df2['para'] != 'DL_string_id-0-0-0']
    df1 = df1[df1['para'] != 'DL_string_id-0-0-0']


    # Round all numeric columns to 4 decimals
    df1 = df1.copy()
    df2 = df2.copy()
    num_cols1 = df1.select_dtypes(include='number').columns
    num_cols2 = df2.select_dtypes(include='number').columns
    df1[num_cols1] = df1[num_cols1].round(4)
    df2[num_cols2] = df2[num_cols2].round(4)

    # Merge row by row (concatenate) and de-duplicate rows
    df3 = pd.concat([df1, df2], ignore_index=True, sort=False).drop_duplicates()
    # Save
    out_path = os.path.join(out_folder, file_name)
    df3.to_csv(out_path, index=False)
```
